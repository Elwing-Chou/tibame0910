- 四大群集
- list: []
    - 清單, 同類型的很多東西
    - key: 0 1 2 3
    - 有序

- dict: {}
    - 字典, 聚集很多欄位, 變成一個複雜單位
    - key是自訂的, "name":"Elwing"
    - 無序

- set: {}
    - 集合, 裡面元素不重複的list
    - key是python內部自己訂的, 你不能用
    - 無序

- tuple: ()
    - 簡化版本的dict
    - 如果我們對位置有共識, 那用tuple就好
    - key 0 1 2, 有序
    - 非常簡化的型態, 查詢操作

```python
c = {"R":255, "G":128, "B":130}
c = (255, 128, 130)
print(type(c))
# 查詢
print(c[0], c[-1])
```
