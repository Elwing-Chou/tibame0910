- 四大群集
- list: []
    - 清單, 同類型的很多東西
    - key: 0 1 2 3
    - 有序

- dict: {}
    - 字典, 聚集很多欄位, 變成一個複雜單位
    - key是自訂的, "name":"Elwing"
    - 無序

- set: {}
    - 集合, 裡面元素不重複的list
    - key是python內部自己訂的, 你不能用
    - 無序

- tuple: ()
    - 簡化版本的dict
    - 如果我們對位置有共識, 那用tuple就好
    - key 0 1 2, 有序
    - 非常簡化的型態, 查詢操作
    - !!!! 不可變動性 -> set裡的東西和字典的key

```python
c = {"R":255, "G":128, "B":130}
c = (255, 128, 130)
print(type(c))
# 查詢
print(c[0], c[-1])
```

- tuple真正重要的意義是他可以當作字典的key或者set裡面的東西
- 什麼不能放在set裡
    - 會變動的(裡面的值可以被修改/可以變成變短): Mutable
        - list, set, dict
    - 不會變動的東西: Unmutable
        - 數字, str, tuple(!!!!)

```python
# main.py
import pygame as pg
import random
import constants as const

#pygame初始化
pg.init()

# --- 字型 ---
def get_font(size):
    for f in ['microsoftjhenghei', 'simhei', 'stheitirelight']:
        if f in pg.font.get_fonts():
            return pg.font.SysFont(f, size)
    return pg.font.SysFont(None, size)

# 字體沒用到(順便給你而已)
FONT_UI = get_font(32)
col, row = const.NUM_COLS, const.NUM_ROWS
inter = const.INTER
width, height = col * inter, row * inter
# 產生視窗
screen = pg.display.set_mode((width, height))
# 設定遊戲標題
pg.display.set_caption("貪食蛇")

snake = []
def reset():
    global snake
    snake = [(0, 2), (0, 1), (0, 0)]

# 把我們邏輯的(i, j) -> UI(x, y)左上角
def ij_to_xy_ul(i, j):
    x = j * inter
    y = i * inter
    return (x, y)

def check_death():
    pass

# 只管蛇的邏輯(狀態): 你絕對不會看到我在update裡做任何畫面的更新
def update():
    pass

# 只管畫面: 你絕對不會看到我在draw裡做任何邏輯的更新
# 更新畫面: 直接把所有東西重畫, 圖層(一層疊一層)
def draw():
    # 準備第一個圖層
    bg = pg.Surface(screen.get_size())
    # 把畫布填滿某個顏色
    bg.fill(const.COLOR_BG)

    # TODO: 畫完棋盤
    # pygame.draw.line(畫布, 顏色, (x坐標1, y坐標1), (x坐標2, y坐標2), 線寬)
    for i in range(col+1):
        pg.draw.line(bg,
                     const.COLOR_LINE,
                     (i * inter, 0),
                     (i * inter, height),
                     1)

    for i in range(row+1):
        pg.draw.line(bg,
                     const.COLOR_LINE,
                     (0, i * inter),
                     (width, i * inter),
                     1)

    # 劃出我的蛇
    for (i, j) in snake:
        x, y = ij_to_xy_ul(i, j)
        pg.draw.rect(bg,
                     const.COLOR_SNAKE_BODY,
                     (x, y, inter, inter),
                     0)
    # 補畫一下頭
    i, j = snake[0]
    x, y = ij_to_xy_ul(i, j)
    pg.draw.rect(bg,
                 const.COLOR_SNAKE_HEAD,
                 (x, y, inter, inter),
                 0)

    # 你要把圖層放到上一層
    screen.blit(bg, (0, 0))
    # 對畫面進行更新(才會真的秀出來)
    pg.display.update()

reset()
draw()
# 1. 建立時鐘物件
clock = pg.time.Clock()
FPS = 10  # 設定每秒最高更新 10 次
# 建立一個永不結束的迴圈(遊戲才不會結束)
running = True
while running:
    # 收取你的遊戲任何事件(滑鼠點擊/鍵盤按鈕...)
    for event in pg.event.get():
        # 偵測鍵盤按下某個按鍵以後放掉的動作
        if event.type == pg.KEYDOWN:
            # 如果需要判斷具體是哪一個：
            if event.key == pg.K_UP:
                pass
            elif event.key == pg.K_DOWN:
                pass
            elif event.key == pg.K_LEFT:
                pass
            elif event.key == pg.K_RIGHT:
                pass
        # 如果收到的事件是按視窗的x
        if event.type == pg.QUIT:
            # 迴圈就會變成while False
            running = False
    update()
    draw()
    if check_death():
        reset()
    # 會幫你算這迴圈在這裡停幾秒再進下一次
    clock.tick(FPS)

pg.quit()
```
