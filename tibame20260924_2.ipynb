```python
import json
import urllib.request as req
import pandas as pd

url = "https://api.gamer.com.tw/anime/v1/danmu.php?videoSn=38250&geo=TW%2CHK"
f = req.urlopen(url)
content = f.read()
content = json.loads(content)
danmus = content["data"]["danmu"]
# pandas: DataFrame表格型態
df = pd.DataFrame(danmus)
df.to_csv("danmu.csv", encoding="utf-8", index=False)
df.to_excel("danmu.xlsx")
df
```

```python
def func(s):
    s = s.replace("#", "")
    r, g, b = s[:2], s[2:4], s[4:]
    r, g, b = int(r, 16), int(g, 16), int(b, 16)
    return (r, g, b)
df["color"].apply(func)
```
